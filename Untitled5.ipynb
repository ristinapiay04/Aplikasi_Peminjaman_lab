```python
# ============================================================
# 1. BUSINESS UNDERSTANDING
# Proyek: Mall Customer Segmentation menggunakan CRISP-DM
# ============================================================

project_name = "Mall Customer Segmentation"
objective = """
Mengelompokkan pelanggan berdasarkan usia, jenis kelamin,
pendapatan tahunan, dan skor belanja menggunakan algoritma
machine learning unsupervised learning (clustering).
"""

print("Nama Proyek:", project_name)
print("\nTujuan Proyek:")
print(objective)

print("\nTahapan CRISP-DM:")
stages = [
    "1. Business Understanding",
    "2. Data Understanding",
    "3. Data Preparation",
    "4. Modeling",
    "5. Evaluation",
    "6. Deployment"
]

for stage in stages:
    print(stage)
```

```python
# ============================================================
# 2. DATA UNDERSTANDING
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()

# Membaca file Excel yang diunggah
file_name = next(iter(uploaded))
df = pd.read_excel(file_name)

print("Ukuran dataset:", df.shape)

print("\n5 Data Pertama:")
display(df.head())

print("\nInformasi Dataset:")
df.info()

print("\nStatistik Deskriptif:")
display(df.describe(include="all"))

print("\nJumlah Missing Values:")
print(df.isnull().sum())

print("\nJumlah Data Duplikat:")
print(df.duplicated().sum())

# Visualisasi distribusi data
df.hist(figsize=(12, 8), bins=20)
plt.suptitle("Distribusi Variabel Numerik")
plt.tight_layout()
plt.show()

# Korelasi variabel numerik
plt.figure(figsize=(8, 5))
sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Korelasi Antarvariabel Numerik")
plt.show()
```

```python
# ============================================================
# 3. DATA PREPARATION
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Salin dataset
data = df.copy()

# Hapus data duplikat
data = data.drop_duplicates()

# Pastikan kolom fitur tersedia
categorical_features = ["Gender"]

numeric_features = [
    "Age",
    "Annual_Income_k$",
    "Spending_Score"
]

required_columns = categorical_features + numeric_features

# Pastikan kolom numerik benar-benar numerik
for col in numeric_features:
    data[col] = pd.to_numeric(data[col], errors="coerce")

# Tangani nilai tidak valid pada fitur
data = data.dropna(subset=required_columns).copy()

# Pisahkan fitur dari identitas pelanggan
X = data[required_columns].copy()

# Pipeline untuk variabel kategorikal
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        drop="first"
    ))
])

# Pipeline untuk variabel numerik
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Gabungkan proses preprocessing
preprocessor = ColumnTransformer([
    ("categorical", categorical_pipeline, categorical_features),
    ("numeric", numeric_pipeline, numeric_features)
])

# Transformasi data
X_processed = preprocessor.fit_transform(X)

# Ubah ke array jika hasilnya sparse
if hasattr(X_processed, "toarray"):
    X_processed = X_processed.toarray()

print("Jumlah data setelah persiapan:", len(data))
print("Jumlah fitur setelah encoding:", X_processed.shape[1])
print("Missing values setelah pembersihan:")
print(data[required_columns].isnull().sum())

print("\nData siap untuk modeling.")
```

```python
# ============================================================
# 4. MODELING
# Eksperimen 6 Algoritma Clustering
# ============================================================

from sklearn.cluster import (
    KMeans,
    AgglomerativeClustering,
    DBSCAN,
    Birch,
    SpectralClustering
)
from sklearn.mixture import GaussianMixture

from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

RANDOM_STATE = 42
N_CLUSTERS = 5

models = {
    "K-Means": KMeans(
        n_clusters=N_CLUSTERS,
        random_state=RANDOM_STATE,
        n_init=10
    ),

    "Agglomerative": AgglomerativeClustering(
        n_clusters=N_CLUSTERS
    ),

    "Gaussian Mixture": GaussianMixture(
        n_components=N_CLUSTERS,
        random_state=RANDOM_STATE
    ),

    "DBSCAN": DBSCAN(
        eps=0.8,
        min_samples=5
    ),

    "Birch": Birch(
        n_clusters=N_CLUSTERS
    ),

    "Spectral Clustering": SpectralClustering(
        n_clusters=N_CLUSTERS,
        random_state=RANDOM_STATE,
        affinity="nearest_neighbors",
        n_neighbors=10
    )
}

results = []
cluster_labels = {}

for name, model in models.items():
    print(f"Melatih model: {name}")

    try:
        # Gaussian Mixture menggunakan fit_predict juga
        labels = model.fit_predict(X_processed)

        cluster_labels[name] = labels

        # Abaikan noise DBSCAN ketika menghitung jumlah cluster
        # untuk menentukan apakah metrik dapat dihitung.
        valid_mask = labels != -1
        valid_labels = labels[valid_mask]
        valid_X = X_processed[valid_mask]

        n_found = len(set(valid_labels))

        # Metrik clustering memerlukan setidaknya 2 cluster
        # dan jumlah sampel lebih banyak daripada cluster.
        if (
            n_found >= 2
            and len(valid_labels) > n_found
        ):
            sil = silhouette_score(valid_X, valid_labels)
            dbi = davies_bouldin_score(valid_X, valid_labels)
            chi = calinski_harabasz_score(valid_X, valid_labels)
        else:
            sil, dbi, chi = np.nan, np.nan, np.nan

        results.append({
            "Algoritma": name,
            "Jumlah Cluster": n_found,
            "Silhouette Score": sil,
            "Davies-Bouldin Index": dbi,
            "Calinski-Harabasz Score": chi
        })

    except Exception as e:
        print("Model gagal:", e)

results_df = pd.DataFrame(results)

print("\nHasil Eksperimen:")
display(
    results_df.sort_values(
        "Silhouette Score",
        ascending=False,
        na_position="last"
    ).reset_index(drop=True)
)
```

```python
# ============================================================
# 5. EVALUATION
# ============================================================

# Metode Elbow dan Silhouette untuk K-Means
k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:
    km = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )

    labels = km.fit_predict(X_processed)

    inertias.append(km.inertia_)
    silhouette_scores.append(
        silhouette_score(X_processed, labels)
    )

# Visualisasi Elbow
plt.figure(figsize=(8, 4))
plt.plot(list(k_values), inertias, marker="o")
plt.title("Metode Elbow - K-Means")
plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Inertia")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()

# Visualisasi Silhouette
plt.figure(figsize=(8, 4))
plt.plot(list(k_values), silhouette_scores, marker="o")
plt.title("Silhouette Score - K-Means")
plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Silhouette Score")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()

# Pilih K dengan Silhouette Score tertinggi
best_k = list(k_values)[np.argmax(silhouette_scores)]

print("Jumlah cluster terbaik menurut Silhouette:", best_k)
print("Silhouette Score terbaik:", max(silhouette_scores))

# Latih ulang K-Means dengan K terbaik
best_kmeans = KMeans(
    n_clusters=best_k,
    random_state=RANDOM_STATE,
    n_init=10
)

data["Cluster"] = best_kmeans.fit_predict(X_processed)

print("\nJumlah pelanggan pada setiap cluster:")
print(data["Cluster"].value_counts().sort_index())

# Profil karakteristik setiap kelompok
print("\nProfil Cluster:")
display(
    data.groupby("Cluster")[
        ["Age", "Annual_Income_k$", "Spending_Score"]
    ].mean().round(2)
)

# Visualisasi pendapatan vs skor belanja
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=data,
    x="Annual_Income_k$",
    y="Spending_Score",
    hue="Cluster",
    palette="tab10",
    s=80
)
plt.title("Segmentasi Pelanggan - K-Means")
plt.xlabel("Pendapatan Tahunan (ribu dolar)")
plt.ylabel("Spending Score")
plt.legend(title="Cluster")
plt.show()

# Perbandingan metrik enam algoritma
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=results_df,
    x="Algoritma",
    y="Silhouette Score",
    ax=axes[0]
)
axes[0].set_title("Perbandingan Silhouette Score")
axes[0].tick_params(axis="x", rotation=65)

sns.barplot(
    data=results_df,
    x="Algoritma",
    y="Davies-Bouldin Index",
    ax=axes[1]
)
axes[1].set_title("Perbandingan Davies-Bouldin Index")
axes[1].tick_params(axis="x", rotation=65)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 6. DEPLOYMENT
# ============================================================

import joblib
from google.colab import files

# Simpan preprocessing dan model
joblib.dump(preprocessor, "preprocessor.pkl")
joblib.dump(best_kmeans, "model_segmentasi.pkl")

print("Model dan preprocessing berhasil disimpan.")

# Contoh pelanggan baru
pelanggan_baru = pd.DataFrame({
    "Gender": ["Female"],
    "Age": [28],
    "Annual_Income_k$": [75],
    "Spending_Score": [82]
})

# Transformasi dengan preprocessing yang sudah dilatih
pelanggan_processed = preprocessor.transform(pelanggan_baru)

if hasattr(pelanggan_processed, "toarray"):
    pelanggan_processed = pelanggan_processed.toarray()

# Prediksi cluster
prediksi = best_kmeans.predict(pelanggan_processed)

print("\nHasil Prediksi Pelanggan Baru:")
print(pelanggan_baru)
print("Pelanggan masuk ke Cluster:", int(prediksi[0]))

# Simpan hasil segmentasi seluruh pelanggan
data.to_csv("hasil_segmentasi_pelanggan.csv", index=False)

print("\nFile hasil segmentasi berhasil dibuat.")

# Unduh hasil segmentasi
files.download("hasil_segmentasi_pelanggan.csv")
```

### **Analisis Ringkasan Profil Statistik untuk Setiap Cluster**

Untuk memahami karakteristik dari masing-masing kelompok pelanggan (Cluster 0, 1, 2, dan 3) yang terbentuk dari hasil pemodelan K-Means, kita akan menghitung beberapa statistik deskriptif utama:

```python
import pandas as pd

# 1. Menghitung profil statistik deskriptif (Mean, Min, Max, Median) untuk setiap cluster
cluster_profile = data.groupby('Cluster')[['Age', 'Annual_Income_k$', 'Spending_Score']].agg(['mean', 'median', 'min', 'max', 'count']).round(2)

print("=== Ringkasan Profil Statistik per Cluster ===")
display(cluster_profile)

# 2. Menghitung distribusi Gender di setiap cluster untuk melengkapi profil
gender_distribution = pd.crosstab(data['Cluster'], data['Gender'], normalize='index') * 100
gender_distribution = gender_distribution.round(2).rename(columns=lambda x: f"Gender_{x} (%)")

print("\n=== Distribusi Persentase Gender per Cluster ===")
display(gender_distribution)
```
